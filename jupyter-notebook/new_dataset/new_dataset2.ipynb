```python
import json
from pathlib import Path

import numpy as np

from pyscf import gto

dict_ = {
    "molecular": [],
    "molecular-g2": [],
    "spin": {},
    "charge": {},
    "reaction-atomic-energy-g2": {},
}

data_path = Path("/home/dhem/workspace/2024.1/data/g2")

with open(data_path / "CHARGE_MULTIPLICITY_W4-11.txt", "r") as f:
    for iter_mol in f:
        mol_data = iter_mol.split(" ")
        i_name = mol_data[0]
        dict_i_name = f"g2-{i_name}"
        dict_["charge"][dict_i_name] = int(mol_data[1])
        dict_["spin"][dict_i_name] = int(mol_data[2]) - 1

        iter_directory = data_path / i_name

        mol = gto.M(
            atom=(iter_directory / "struc.xyz").resolve().as_posix(),
            basis="sto-3g",
            verbose=0,
            charge=dict_["charge"][dict_i_name],
            spin=dict_["spin"][dict_i_name],
            unit="B",
        )
        mol.build()

        molecular = []
        for i_atom in mol._atom:
            molecular.append(
                [
                    i_atom[0],
                    i_atom[1][0],
                    i_atom[1][1],
                    i_atom[1][2],
                ]
            )
        dict_[dict_i_name] = molecular
        dict_["molecular"].append(dict_i_name)
        dict_["molecular-g2"].append(dict_i_name)

with open(data_path / ".res", "r") as f:
    for iter_mol in f:
        # if not begin with tmer2_GMTKN
        if not iter_mol.startswith("tmer2_GMTKN"):
            continue
        mol_data = iter_mol.split(" ")
        systems_list = []
        stoichiometry_list = []
        for iter_system in mol_data[1:]:
            if iter_system == "x":
                break
            systems_list.append(f"g2-{iter_system}")

        for iter_stoichiometry in mol_data[
            2 + len(systems_list) : 2 + 2 * len(systems_list)
        ]:
            stoichiometry_list.append(int(iter_stoichiometry))
        dict_["reaction-atomic-energy-g2"][f"g2-{mol_data[1]}"] = {
            "systems": systems_list,
            "stoichiometry": stoichiometry_list,
        }

for number_atom in range(200):
    dict_[f"molecular{number_atom}"] = []
    print(f"number_atom = {number_atom}")
    for molecular in dict_["molecular"]:
        num_heavy_atom = 0
        num_atom = 0
        for atom in dict_[molecular]:
            num_atom += 1
            if atom[0] != "H":
                num_heavy_atom += 1
        if number_atom == 0:
            if num_atom == 1:
                print(f'"{molecular}",')
                dict_[f"molecular{number_atom}"].append(molecular)
        else:
            if num_atom != 1 and num_heavy_atom == number_atom:
                print(f'"{molecular}",')
                dict_[f"molecular{number_atom}"].append(molecular)
    print()
print("DONE")
```

```python
dict_new = {}
for key in dict_:
    if key.startswith("molecular"):
        if len(dict_[key]) > 0:
            dict_new[key] = dict_[key]
for key in dict_:
    if not key.startswith("molecular"):
        dict_new[key] = dict_[key]

with open("../cadft/utils/g2.json", "w") as f:
    json.dump(dict_new, f)
```
