```python
from ase.collections import g2

import json
import numpy as np

from pyscf.data.elements import ELEMENTS

dict_ = {
    "molecular": [],
    "molecular_g2-ase": [],
    "spin": {},
    "charge": {},
    "reaction-atomic-energy": {},
}
name = g2.names

for i_name in name:
    positions = np.array(g2[i_name].get_positions())
    numbers = g2[i_name].numbers

    # print(numbers)

    molecular = []
    total_ele = 0
    for i_number, atom_number in enumerate(numbers):
        atom = []
        total_ele += atom_number
        atom.append(ELEMENTS[atom_number])
        atom.append(float(positions[i_number][0]))
        atom.append(float(positions[i_number][1]))
        atom.append(float(positions[i_number][2]))
        molecular.append(atom)

    dict_i_name = f"g2-ase-{i_name}"
    dict_[dict_i_name] = molecular
    dict_["molecular_g2-ase"].append(dict_i_name)
    dict_["molecular"].append(dict_i_name)
    dict_["spin"][dict_i_name] = int(sum(g2[i_name].get_initial_magnetic_moments()))
    dict_["charge"][dict_i_name] = int(sum(g2[i_name].get_initial_charges()))

    systems_list = [dict_i_name]
    stoichiometry_list = ["-1"]
    for atom_number in numbers:
        systems_list.append(f"g2-ase-{ELEMENTS[atom_number]}")
        stoichiometry_list.append("1")

    dict_["reaction-atomic-energy"][dict_i_name] = {
        "systems": systems_list,
        "stoichiometry": stoichiometry_list,
    }

# sort the molecular
dict_["molecular"].sort()
dict_["molecular_g2-ase"].sort()
dict_["spin"] = dict(sorted(dict_["spin"].items()))
dict_["charge"] = dict(sorted(dict_["charge"].items()))
dict_["reaction-atomic-energy-g2-ase"] = dict(
    sorted(dict_["reaction-atomic-energy"].items())
)

for number_atom in range(100):
    dict_[f"molecular{number_atom}"] = []
    print(f"number_atom = {number_atom}")
    for molecular in dict_["molecular"]:
        num_heavy_atom = 0
        num_atom = 0
        for atom in dict_[molecular]:
            num_atom += 1
            if atom[0] != "H":
                num_heavy_atom += 1
        if number_atom == 0:
            if num_atom == 1:
                print(f'"{molecular}",')
                dict_[f"molecular{number_atom}"].append(molecular)
        else:
            if num_atom != 1 and num_heavy_atom == number_atom:
                print(f'"{molecular}",')
                dict_[f"molecular{number_atom}"].append(molecular)
    print()
```

```python
print(dict_)

dict_new = {}
for key in dict_:
    if key.startswith("molecular"):
        if dict_[key]:
            dict_new[key] = dict_[key]
for key in dict_:
    if not key.startswith("molecular"):
        dict_new[key] = dict_[key]

print(dict_new)

with open(f"../cadft/utils/g2_ase.json", "w") as f:
    json.dump(dict_new, f)
```
