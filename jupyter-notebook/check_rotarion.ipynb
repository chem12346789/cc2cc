```python
import pandas as pd
import numpy as np

data_path_list = [
    "/home/chenzihao/workspace/cc2cc_test5/validate/ccdft_cc-pVDZ_atom-1-1513512_gmtkn-cc-pVDZ.csv",
    "/home/chenzihao/workspace/cc2cc_test5/validate/ccdft_cc-pVDZ_atom-1-3036943_gmtkn-cc-pVDZ.csv",
    "/home/chenzihao/workspace/cc2cc_test5/validate/ccdft_cc-pVDZ_atom-1-1428227_gmtkn-cc-pVDZ.csv",
    "/home/chenzihao/workspace/cc2cc_test5/validate/ccdft_cc-pVDZ_atom-1-4149292_gmtkn-cc-pVDZ.csv",
    # "/home/chenzihao/workspace/cc2cc_test5/validate/ccdft_cc-pVDZ_atom-1-1235076_gmtkn-cc-pVDZ.csv",
    # "/home/chenzihao/workspace/cc2cc_test5/validate/ccdft_cc-pVDZ_atom-1-4190857_gmtkn-cc-pVDZ.csv",
    # "/home/chenzihao/workspace/cc2cc_test5/validate/ccdft_cc-pVDZ_atom-1-1256130_gmtkn-cc-pVDZ.csv",
]


for data_path in data_path_list:
    data_record = np.zeros((5, 2))
    stoichiometry_list = np.array(
        [
            [-5, 5, -1, 0, 0],
            [-4, 5, 0, -1, 0],
            [-18, 16, 0, 0, -1],
        ]
    )
    data = pd.read_csv(data_path)
    for i in range(5):
        if len(data) > (i * 100):
            data_scf = data["delta_scf"].to_numpy()[i * 100 : i * 100 + 100]
            delta_dft = data["delta_dft"].to_numpy()[i * 100 : i * 100 + 100]
            data_scf_min = np.min(data_scf)
            data_scf_max = np.max(data_scf)
            data_dft_min = np.min(delta_dft)
            data_dft_max = np.max(delta_dft)
            data_record[i, 0] = np.mean(data_scf)
            data_record[i, 1] = np.mean(delta_dft)
            print(data_scf_max - data_scf_min, data_dft_max - data_dft_min)
        else:
            print("NAN")
    print("")
    print(np.einsum("ij,ki->jk", data_record, stoichiometry_list))
```
