```python
import sys

sys.path.append("/home/chenzihao/workspace/cc2cc_atom/")

import numpy as np
from pyscf import gto, scf, dft

from cc2cc.utils import gen_mole

dis_list = np.linspace(-0.5, 0.5, 11)
ene_list = np.zeros(len(dis_list))
for i, dis in enumerate(dis_list):
    mol, name = gen_mole(
        "ch3",
        "0-1",
        0,
        dis,
        "cc-pvdz",
        True,  # Changed "True" to True (boolean)
        "g2",
        verbose=0,
    )

    if mol.spin == 0:
        mdft = dft.RKS(mol)
    else:
        mdft = dft.UKS(mol)
    mdft.xc = "b3lyp"
    mdft.verbose = 1
    mdft.kernel()
    print(f"Total energy: {mdft.e_tot}")
    ene_list[i] = mdft.e_tot
```

```python
import matplotlib.pyplot as plt

plt.plot(dis_list, ene_list, "o-")
```
