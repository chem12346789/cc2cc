```python
import py3Dmol

from cc2cc.utils import gen_mole

name_mol_list = [
    "SIE4x4-he+",
    # "Amino20x4-MET_xag",
    # "Amino20x4-MET_xav",
    # "Amino20x4-MET_xbf",
    # "Amino20x4-MET_xbm",
    # "Amino20x4-MET_xbo",
]

for name_mol in name_mol_list:
    # Generate the molecule with the specified parameters
    # The parameters are:
    #       name, stratch_atom, stratch_orientation, stratch_distance and basis set
    mol = gen_mole(name_mol, 0, 1, 0, "cc-pVDZ", True, "gmtkn")

    xyz = mol.tostring(format="xyz")

    xyzview = py3Dmol.view()
    xyzview.addModel(xyz, "xyz")
    xyzview.setStyle({"stick": {}, "sphere": {"scale": 0.3}})
    xyzview.setStyle(
        {
            "stick": {},
            "sphere": {"scale": 0.3},
            "label": "atom",
        }
    )
    xyzview.setHoverable(
        {},
        True,
        """function(atom,viewer,event,container) {
            if(!atom.label) {
            atom.label = viewer.addLabel(atom.atom,{position: atom, backgroundColor: 'mintcream', fontColor:'black', fontSize: 18, fontWeight: 'bold', alignment: 'center', showBackground: true, showBorder: true, borderColor: 'black', borderWidth: 1, padding: 2, zIndex: 1000});
            }}""",
        """function(atom,viewer) {
            if(atom.label) {
            viewer.removeLabel(atom.label);
            delete atom.label;
            }
        }""",
        viewer=(0, 0),
    )
    xyzview.zoomTo()
    xyzview.show()
```
