```python
import json
from pathlib import Path
from datetime import datetime

import pandas as pd
import numpy as np

date = datetime.now().strftime("%Y-%m-%d")

verbose = 1
data_set = "gmtkn-cc-pVDZ"

data_path_list = sorted(
    list(Path("../validate").glob(f"*gmtkn*.csv")),
    key=lambda p: p.stat().st_ctime,
)

real_data_path = ("/home/chenzihao/workspace/cc2cc_test5/validate/backup/ccdft_cc-pVDZ_atom-1-test_gmtkn-cc-pVDZ.csv"
)
real_data = pd.read_csv(real_data_path)


for data_path in data_path_list:
    data = pd.read_csv(data_path)
    for name in data["name"]:
        if name in real_data["name"].values:
            real_row = real_data[real_data["name"] == name]
            data_row = data[data["name"] == name]
            if not np.isclose(real_row["delta_d3bj"].values[0], data_row["delta_d3bj"].values[0]):
                print(f"Warning: {name} delta_d3bj mismatch: {real_row['delta_d3bj'].values[0]} vs {data_row['delta_d3bj'].values[0]}")
                data.loc[data["name"] == name, "delta_d3bj"] = real_row[
                    "delta_d3bj"
                ].values[0]
            if not np.isclose(real_row["delta_d3zero"].values[0], data_row["delta_d3zero"].values[0]):
                print(f"Warning: {name} delta_d3zero mismatch: {real_row['delta_d3zero'].values[0]} vs {data_row['delta_d3zero'].values[0]}")
                data.loc[data["name"] == name, "delta_d3zero"] = real_row[
                    "delta_d3zero"
                ].values[0]

    # save the processed data
    data.to_csv(data_path, index=False)
```
